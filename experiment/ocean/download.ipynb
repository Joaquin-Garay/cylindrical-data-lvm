# NDBC station 46042 standard meteorological data

Download NOAA/NDBC historical standard meteorological files for 2000-2022 and load them into one cleaned dataframe.

```python
import sys
from pathlib import Path
import experiment.ocean.helpers as mod
import pandas as pd

cwd = Path.cwd().resolve()
candidate_project_roots = [cwd, *cwd.parents]
```

```python
PROJECT_ROOT = mod.find_project_root()
DATA_DIR = PROJECT_ROOT / "data" / "ndbc"
```

```python
file_paths = [mod.download_ndbc_stdmet_file(mod.STATION, year) for year in mod.YEARS]

ndbc_df = pd.concat(
    [
        mod.read_ndbc_stdmet_file(path, station=mod.STATION, file_year=year)
        for path, year in zip(file_paths, mod.YEARS)
    ],
    ignore_index=True,
)
ndbc_df = ndbc_df.sort_values("time").reset_index(drop=True)
ndbc_df.to_csv(DATA_DIR / "ocean_data.csv", index=False)
ndbc_df
```
